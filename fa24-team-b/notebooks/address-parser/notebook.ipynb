```python
import pandas as pd
from retraining import Retraining
import random
```

# Train

```python
random.seed(42)
retrainer = Retraining(path_to_train='./data/train.csv')
retrainer.retrain_model(epochs=2)
model = retrainer.model
```

# Test

```python
# test the parser
from deepparse.dataset_container import ListDatasetContainer

addresses = [
    ('13-a main st 4', ['StreetNumber', 'StreetName', 'StreetName', 'Unit']),
    ('13a main st unit 4', ['StreetNumber', 'StreetName', 'StreetName', 'Unit', 'Unit']),
    ('13 a main st #4', ['StreetNumber', 'StreetName', 'StreetName', 'StreetName', 'Unit']),
    ('11 w broadway street apt 607', ['StreetNumber', 'StreetName', 'StreetName', 'StreetName', 'StreetName', 'Unit']),
    ('95 saint alphonsus st 203', ['StreetNumber', 'StreetName', 'StreetName', 'StreetName', 'Unit']),
]

model.test(ListDatasetContainer(addresses))
```

```python
# let's see how it parses them now
for address, _ in addresses:
    print(model(address))
```
