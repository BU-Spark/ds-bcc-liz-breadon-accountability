```python
import pandas as pd

# Path to the Excel file
file_path = r'C:\Users\sul19\Desktop\701 Project\Property Assessment Datasets\Final Excels\Merged_Property_Assessment V2.xlsx'
output_excel_path = r'C:\Users\sul19\Desktop\701 Project\Property Assessment Datasets\Final Excels\Merged_Property_Assessment_V3_Scores_FullAddresses.xlsx'
output_csv_path = r'C:\Users\sul19\Desktop\701 Project\Property Assessment Datasets\Final Excels\Merged_Property_Assessment_V3_Scores_FullAddresses.csv'

# Load the Excel file
df = pd.read_excel(file_path)

# Create the Full Address column by combining ST_NUM and ST_NAME, with a condition for empty ST_NUM
df['Full Address'] = df.apply(
    lambda row: f"{row['ST_NUM']} {row['ST_NAME']}" if pd.notna(row['ST_NUM']) else row['ST_NAME'],
    axis=1
)

# Add Property_Condition_Score based on the overall_cond_label column
condition_score_map = {
    'Unsound': 1,
    'Fair': 2,
    'Average': 3,
    'Good': 4,
    'Very Good': 5
}
df['Property_Condition_Score'] = df['overall_cond_label'].map(condition_score_map)

# Add Building_Classification_Score based on the building_classification column
classification_score_map = {
    'Unknown': 0,
    'Old': 1,
    'Average': 2,
    'New': 3
}
df['Building_Classification_Score'] = df['building_classification'].map(classification_score_map)

# Save the modified dataframe to a new Excel and CSV file
df.to_excel(output_excel_path, index=False)
df.to_csv(output_csv_path, index=False)
```
