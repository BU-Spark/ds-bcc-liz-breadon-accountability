```python
import pandas as pd

# Load both datasets
student_df = pd.read_excel('StudentAddresses-2016-2024.xlsx')
sam_df = pd.read_csv('Live_Street_Address_Management_(SAM)_Addresses.csv')

# Add source column
student_df['Source'] = 'Student'
sam_df['Source'] = 'SAM'

# Define abbreviation mapping
abbreviation_mapping = {
    'ST': 'STREET',
    'RD': 'ROAD',
    'AVE': 'AVENUE',
    'BLVD': 'BOULEVARD',
    'DR': 'DRIVE',
    'LN': 'LANE',
    'PL': 'PLACE',
    'CT': 'COURT',
    'TER': 'TERRACE',
    'CIR': 'CIRCLE',
    # Add other mappings as needed
}

# Function to expand abbreviations
def expand_abbreviations(address, mapping):
    words = address.split()
    expanded_words = [mapping.get(word, word) for word in words]
    return ' '.join(expanded_words)

# Create FULL_ADDRESS columns
student_df['FULL_ADDRESS'] = (
    student_df['6a. street #'].astype(str).str.strip().str.upper() + ' ' +
    student_df['6b. street name'].astype(str).str.strip().str.upper() + ' ' +
    student_df['6c. street suffix'].astype(str).str.strip().str.upper() + ', Boston, MA'
).apply(lambda x: expand_abbreviations(x, abbreviation_mapping))

sam_df['FULL_ADDRESS'] = (
    sam_df['STREET_NUMBER_SORT'].astype(str).str.strip().str.upper() + ' ' +
    sam_df['STREET_BODY'].astype(str).str.strip().str.upper() + ' ' +
    sam_df['STREET_FULL_SUFFIX'].astype(str).str.strip().str.upper() + ', Boston, MA'
)

# Remove duplicates **within each dataset** before combining
unique_student_addresses = student_df[['FULL_ADDRESS', 'Source']].drop_duplicates(subset=['FULL_ADDRESS'])
unique_sam_addresses = sam_df[['FULL_ADDRESS', 'Source']].drop_duplicates(subset=['FULL_ADDRESS'])

# Print for debugging
print(unique_sam_addresses)
print(unique_student_addresses)
```

```python
# Combine datasets while maintaining dataset-level uniqueness
combined_df = pd.concat([unique_student_addresses, unique_sam_addresses])

# Drop duplicates between the two datasets
combined_df = combined_df.drop_duplicates(subset=['FULL_ADDRESS'])

# Split into batches of 10,000
batch_size = 10000
batches = [combined_df.iloc[i:i + batch_size] for i in range(0, len(combined_df), batch_size)]

# Save each batch to a CSV
for idx, batch in enumerate(batches):
    batch.to_csv(f'Batch_{idx + 1}.csv', index=False)

print(f"Data split into {len(batches)} batches and saved successfully.")
```

```python
import pandas as pd
import requests
import time

# Function to geocode addresses using US Census Geocoder
def geocode_census(address):
    try:
        base_url = "https://geocoding.geo.census.gov/geocoder/locations/onelineaddress"
        params = {
            'address': address,
            'benchmark': 'Public_AR_Current',
            'format': 'json'
        }
        response = requests.get(base_url, params=params)
        if response.status_code == 200:
            result = response.json()
            if result.get('result') and result['result'].get('addressMatches'):
                location = result['result']['addressMatches'][0]['coordinates']
                return location['x'], location['y']  # Longitude, Latitude
        return None, None
    except Exception as e:
        print(f"Error geocoding {address}: {e}")
        return None, None

# Process a single batch
def process_batch(batch_file, output_file):
    # Read the batch file
    batch_df = pd.read_csv(batch_file)
    
    # Prepare output columns
    batch_df['Longitude'] = None
    batch_df['Latitude'] = None

    # Counter for failed geocoding
    failed_count = 0

    # Geocode each address
    for index, row in batch_df.iterrows():
        address = row['FULL_ADDRESS']
        print(f"Processing address {index + 1}/{len(batch_df)}: {address}")
        longitude, latitude = geocode_census(address)
        if longitude is not None and latitude is not None:
            print(f"Success: {address} -> Longitude: {longitude}, Latitude: {latitude}")
        else:
            print(f"Failed: {address}")
            failed_count += 1
        batch_df.at[index, 'Longitude'] = longitude
        batch_df.at[index, 'Latitude'] = latitude
        time.sleep(0.2)  # To respect rate limits

    # Save the geocoded batch
    batch_df.to_csv(output_file, index=False)
    print(f"Geocoded batch saved to {output_file}")
    
    # Print the number of failed addresses
    print(f"Number of addresses that failed to geocode: {failed_count}")

# Example usage
batch_number = 1  # Replace with the appropriate batch number
input_file = f'Batch_{batch_number}.csv'  # Adjust the file path if needed
output_file = f'Geocoded_Batch_{batch_number}.csv'
process_batch(input_file, output_file)
```
