```python
data = pd.read_csv('combined_file.csv')
data.head(15)
```

```python
print("Column Names and Data Types:")
print(data.dtypes)
```

```python
# Check if there are rows where `submitted_photo` or `closed_photo` are not NaN
non_nan_rows = data[(~data['submitted_photo'].isna()) | (~data['closed_photo'].isna())]

# Display the result
print(non_nan_rows)

# Count the number of non-NaN rows
print(f"Number of rows where either column is not NaN: {len(non_nan_rows)}")
```

```python
combined_df = pd.concat(
    (pd.read_csv(file, dtype={'submitted_photo': str, 'closed_photo': str}) for file in all_files),
    ignore_index=True
)
```

```python
combined_df['submitted_photo'] = combined_df['submitted_photo'].astype(str)
combined_df['closed_photo'] = combined_df['closed_photo'].astype(str)
```

```python
# Save the DataFrame to a new CSV file
combined_df.to_csv("combined_cleaned_data.csv", index=False)
```

```python
data = pd.read_csv('combined_file.csv')
data.head(15)
```

```python
# Ensure the "open_dt" column is treated as a string (in case it isn't already)
data['open_dt'] = data['open_dt'].astype(str)

# Extract the first four characters from the "open_dt" column and create a new column "Year"
data['Year'] = data['open_dt'].str[:4]

# Check the updated DataFrame
print(data[['open_dt', 'Year']].head())
```
