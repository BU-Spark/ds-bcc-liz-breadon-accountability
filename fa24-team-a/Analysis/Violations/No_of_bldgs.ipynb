```python
import pandas as pd
import numpy as np
```

```python
sam_adds = pd.read_csv("Geolocated_SAMAddresses.csv")
```

```python
sam_adds.info()
```

```python
n_zip = pd.read_csv("Updated_Zip_Code_and_Neighborhood_Mapping.csv")
n_zip['Zip Code'] = n_zip['Zip Code'].astype(str)
n_zip['Zip Code'] = n_zip['Zip Code'].apply(lambda x: '0'+x) 
```

```python
sam_adds['ZIP_CODE'] = sam_adds['ZIP_CODE'].astype(str)
sam_adds['ZIP_CODE'] = sam_adds['ZIP_CODE'].apply(lambda x: "0"+x)
```

```python
sam_adds = pd.merge(sam_adds,n_zip, left_on='ZIP_CODE', right_on='Zip Code', how='left')
```

```python
sam_adds_bldgs = sam_adds[['BUILDING_ID','Zip Code','Neighbourhood']]
sam_adds_bldgs = sam_adds_bldgs.drop_duplicates()
```

```python
no_of_bldgs = sam_adds_bldgs.groupby(['Neighbourhood']).size()
```

```python
no_of_bldgs =no_of_bldgs.reset_index()
no_of_bldgs.columns = ['Neighbourhood','no_of_bldgs']
```

```python
no_of_bldgs.sort_values(by='no_of_bldgs', ascending=False, inplace=True)
```

```python
no_of_bldgs.head()
```

```python
student_sam = pd.read_csv('Final_dataset.csv')
```

```python
student_sam['Neighbourhood']
```

```python
len(student_sam.sam_id.unique())
```

```python
no_of_bldgs_student=student_sam[['Neighbourhood','sam_id']].drop_duplicates('sam_id').groupby('Neighbourhood').size()
```

```python
no_of_bldgs_student=no_of_bldgs_student.reset_index()
no_of_bldgs_student.columns = ['Neighbourhood','no_of_bldgs_student']
```

```python
no_of_bldgs_student
```

```python
no_of_bldgs = pd.merge(no_of_bldgs, no_of_bldgs_student, on='Neighbourhood', how='left')
```

```python
no_of_bldgs.dropna(inplace=True)
```

```python
no_of_bldgs['%_bldgs_occupied_by_students'] = no_of_bldgs['no_of_bldgs_student']*100/no_of_bldgs['no_of_bldgs']
```

```python
no_of_bldgs
```

```python
no_of_bldgs=pd.melt(
    no_of_bldgs,
    id_vars=["Neighbourhood", "%_bldgs_occupied_by_students"],
    value_vars=["no_of_bldgs_student", "no_of_bldgs"],
    var_name="Building Type",
    value_name="Count"
)
```

```python
no_of_bldgs
```

```python
no_of_bldgs.to_csv('No_of_bldgs.csv',index=False)
```
