Converting timestamp to datetime format

```python
n_zip =pd.read_csv('Updated_Zip_Code_and_Neighborhood_Mapping.csv')
```

```python
n_zip['Zip Code'] = n_zip['Zip Code'].astype(str)
```

```python
n_zip['Zip Code'] = n_zip['Zip Code'].apply(lambda x: '0' +x)
```

```python
df = pd.merge(df,n_zip, left_on='violation_zip', right_on='Zip Code', how='left')
```

```python
df
```

Calculating number of violations every year for each unit

```python
df_neigh_year = df.groupby(['Neighbourhood','year','sam_id']).size()
```

```python
df_neigh_year = df_neigh_year.reset_index()
df_neigh_year.columns = ['Neighbourhood','Year','sam_id','Count_of_violations']
```

```python
df_neigh_year
```

```python
len(df_neigh_year.sam_id.unique())
```

Calculating number of bad landlords

```python
def violations_in_prev_year(row):
    year = row['Year']
    sam = row['sam_id']
    
    prev_violations = df_neigh_year[(df_neigh_year['Year']==year-1) & (df_neigh_year['sam_id']==sam)]['Count_of_violations'].sum()
    try:
        prev_violations = df_neigh_year[(df_neigh_year['Year']==year-1) & (df_neigh_year['sam_id']==sam)]['Count_of_violations'].sum()
    except:
        prev_violations = np.nan

    return prev_violations
```

```python
temp = df_neigh_year.copy(deep=True)
```

```python
len(temp.apply(violations_in_prev_year,axis=1))
```

```python
temp['prev_year_violation'] = temp.apply(violations_in_prev_year,axis=1)
```

```python
temp[(temp['Count_of_violations']>5) & (temp['sam_id']!=0)]
```

```python
temp[(temp['prev_year_violation']>5) & (temp['sam_id']!=0)]
```

## Number of Buildings

```python
sam_adds = pd.read_csv("Geolocated_SAMAddresses.csv")
```

```python
sam_adds.info()
```

```python
n_zip
```

```python
sam_adds['ZIP_CODE'] = sam_adds['ZIP_CODE'].astype(str)
sam_adds['ZIP_CODE'] = sam_adds['ZIP_CODE'].apply(lambda x: "0"+x)
```

```python
sam_adds = df = pd.merge(sam_adds,n_zip, left_on='ZIP_CODE', right_on='Zip Code', how='left')
```

```python
sam_adds_bldgs = sam_adds[['BUILDING_ID','Zip Code','Neighbourhood']]
```

```python
sam_adds_bldgs = sam_adds_bldgs.drop_duplicates()
```

```python
no_of_bldgs = sam_adds_bldgs.groupby(['Neighbourhood']).size()
```

```python
no_of_bldgs =no_of_bldgs.reset_index()
no_of_bldgs.columns = ['Neighbourhood','no_of_bldgs']
```

```python
no_of_bldgs
```

```python
df.info()
```

```python
census = pd.read_csv('census.csv')
```

```python
census.info()
```

```python
no_of_bldgs = pd.merge(no_of_bldgs, census[['Neighbourhood','Total:.3']], left_on='Neighbourhood', right_on='Neighbourhood', how='left')
```

```python
no_of_bldgs.columns = ['Neighbourhood', 'no_of_bldgs_student', 'no_of_bldgs_census']
```

```python
no_of_bldgs.dropna(inplace=True)
```

```python
no_of_bldgs['% of student occupied buildings'] = no_of_bldgs['no_of_bldgs_student']*100/no_of_bldgs['no_of_bldgs_census']
```

Calculating number of bldgs occupied by students versus total number of buildings

```python
no_of_bldgs
```

```python
no_of_bldgs = pd.melt(
    no_of_bldgs,
    id_vars=["Neighbourhood", "% of student occupied buildings"],
    value_vars=["no_of_bldgs_student", "no_of_bldgs_census"],
    var_name="Building Type",
    value_name="Count"
)
```

```python
no_of_bldgs.to_csv('No_of_bldgs.csv',index=False)
```
